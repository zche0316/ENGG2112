# Final Model Comparison

## 1. This cell and bar chart indicates the trend of fire numbers based on three different kinds of weather data.

```python
import matplotlib.pyplot as plt # type: ignore
import numpy as np # type: ignore

weather_types = ['Temperature', 'Wind', 'Humidity']
r2_values = [0.160682, 0.195110, 0.489728] 

plt.figure(figsize=(10, 6))
bars = plt.bar(weather_types, r2_values)

plt.title('Comparison of the trend of fire numbers based on weather data (Based on R² values)', fontsize=14)
plt.xlabel('Weather condition data', fontsize=12)
plt.ylabel('R² value', fontsize=12)

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.3f}',
             ha='center', va='bottom')

plt.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()

plt.show()
```
